```python
bad_urls = []

for i, url in enumerate(df["url"]):
    try:
        urlparse(url)
    except ValueError:
        bad_urls.append((i, url))

print(f"Found {len(bad_urls)} malformed URLs.")
print(bad_urls[:10])  # Show the first 10
```

```python
# Common phishing-related keywords
SUSPICIOUS_WORDS = [
    "login", "signin", "verify", "account", "update",
    "secure", "bank", "paypal", "confirm", "password",
    "webscr", "ebay", "wp", "admin"
]


def shannon_entropy(text):
    """Calculate Shannon entropy of a string."""
    if not text:
        return 0

    entropy = 0
    for c in set(text):
        p = text.count(c) / len(text)
        entropy -= p * math.log2(p)
    return entropy


def extract_features(url):
    """
    Extract URL-based features for phishing detection.
    Returns a dictionary.
    """
    if not isinstance(url, str):
        return {}

    url = url.replace("[.]", ".")
    parsed = urlparse(url)

    domain = parsed.netloc
    path = parsed.path
    query = parsed.query

    features = {}

    # -----------------------
    # Basic lengths
    # -----------------------
    features["url_length"] = len(url)
    features["domain_length"] = len(domain)
    features["path_length"] = len(path)
    features["query_length"] = len(query)

    # -----------------------
    # Character counts
    # -----------------------
    features["num_dots"] = url.count(".")
    features["num_hyphens"] = url.count("-")
    features["num_underscores"] = url.count("_")
    features["num_slashes"] = url.count("/")
    features["num_questionmarks"] = url.count("?")
    features["num_equal"] = url.count("=")
    features["num_at"] = url.count("@")
    features["num_ampersand"] = url.count("&")
    features["num_percent"] = url.count("%")
    features["num_digits"] = sum(c.isdigit() for c in url)

    # -----------------------
    # Binary features
    # -----------------------
    features["has_https"] = int(parsed.scheme == "https")
    features["has_ip"] = int(bool(re.match(r"^\d+\.\d+\.\d+\.\d+$", domain)))
    features["has_port"] = int(":" in domain)

    # -----------------------
    # Subdomains
    # -----------------------
    features["num_subdomains"] = max(domain.count(".") - 1, 0)

    # -----------------------
    # Suspicious words
    # -----------------------
    lower_url = url.lower()

    for word in SUSPICIOUS_WORDS:
        features[f"contains_{word}"] = int(word in lower_url)

    # -----------------------
    # Entropy
    # -----------------------
    features["entropy"] = shannon_entropy(url)

    return features
```

| Feature           | What it captures                                     | Why it may indicate phishing                                                     |
| ----------------- | ---------------------------------------------------- | -------------------------------------------------------------------------------- |
| URL length        | Is the URL unusually long?                           | Long URLs can hide the real destination.                                         |
| Domain length     | Is the hostname unusually long?                      | Fake domains often include many keywords.                                        |
| Path/query length | Is there an unusually complex path or query?         | Auto-generated phishing pages and redirects often have long paths or parameters. |
| Dots/subdomains   | Is the hostname deeply nested?                       | Multiple subdomains can make a fake site look legitimate.                        |
| Hyphens           | Does the domain stitch together many words?          | Common in impersonation domains.                                                 |
| Digits            | Are there many numbers?                              | Random IDs are common in generated phishing URLs.                                |
| `@` symbol        | Does the URL hide the real destination?              | A classic phishing trick.                                                        |
| IP address        | Is an IP used instead of a domain?                   | Often associated with suspicious or temporary sites.                             |
| HTTPS             | Is the connection encrypted?                         | Weak signal today, but still adds context.                                       |
| Suspicious words  | Does the URL contain terms like "login" or "verify"? | These are frequently used to impersonate trusted services.                       |
| Entropy           | Does the URL look random?                            | High randomness often reflects generated strings or obfuscation.                 |


```python
features_df = df["url"].apply(extract_features).apply(pd.Series)
```

```python
new_df = pd.concat(
    [df[["url", "label"]], features_df],
    axis=1
)
```
